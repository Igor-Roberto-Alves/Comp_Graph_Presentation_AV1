```python
import torch
import gc

def reset_gpu_memory():
    # 1. Deleta variáveis que podem estar segurando referências
    gc.collect()
    
    # 2. Limpa o cache de memória não utilizada que o PyTorch reserva na GPU
    torch.cuda.empty_cache()
    
    # 3. (Opcional) Reseta as estatísticas de pico para monitoramento limpo
    torch.cuda.reset_peak_memory_stats()
```
